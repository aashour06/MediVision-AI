## Categorical Features

```python
df["patient_gender"] = df["patient_gender"].astype("category")
df["view_position"] = df["view_position"].astype("category")
```

## Finding Labels

```python
labels = df["finding_labels"].str.split("|").explode()

labels.value_counts()
```

```python
unique_labels = sorted(labels.unique())

print("Number of unique labels:", len(unique_labels))
print(unique_labels)
```

```python
no_finding_mixed = df[
    df["finding_labels"].str.contains(
        "No Finding",
        na=False
    ) &
    df["finding_labels"].str.contains(
        r"\|",
        regex=True,
        na=False
    )
]

len(no_finding_mixed)
```

```python
df["num_diseases"] = df["finding_labels"].str.count(r"\|") + 1

df["num_diseases"].value_counts().sort_index()
```

```python
unique_labels = set(labels.unique())

for label in unique_labels:
    if label != "No Finding":
        df[label] = df["finding_labels"].apply(lambda x: 1 if label in x.split("|") else 0)

df["has_finding"] = (df["finding_labels"] != "No Finding").astype(int)
```

## Image Metadata Sanity Check

```python
invalid_dimensions = df[
    (df["image_width"] <= 0) | 
    (df["image_height"] <= 0) | 
    (df["pixel_spacing_x"] <= 0) | 
    (df["pixel_spacing_y"] <= 0)
]

len(invalid_dimensions)
```

```python
(
    df["pixel_spacing_x"] ==
    df["pixel_spacing_y"]
).value_counts()
```

## Patient metadata consistency

```python
gender_consistency = (
    df.groupby("patient_id")["patient_gender"]
    .nunique()
)

gender_consistency.value_counts()
```

```python
age_stats = (
    df.groupby("patient_id")["patient_age"]
    .agg(["min", "max", "nunique"])
)

age_stats.sort_values(
    "nunique",
    ascending=False
).head(20)
```

```python
images_per_patient = (
    df.groupby("patient_id")
    .size()
)

images_per_patient.describe()
```

```python
images_per_patient.sort_values(
    ascending=False
).head(20)
```

```python
quality_report = pd.DataFrame({
    "metric": [
        "Number of rows",
        "Number of columns",
        "Unique images",
        "Unique patients",
        "Missing values",
        "Duplicate rows",
        "Invalid ages",
        "Invalid dimensions",
        "Unique findings",
        "PA images",
        "AP images"
    ],
    
    "value": [
        len(df),
        df.shape[1],
        df["image_id"].nunique(),
        df["patient_id"].nunique(),
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        (
            (df["patient_age"] < 0) |
            (df["patient_age"] > 100)
        ).sum(),
        len(invalid_dimensions),
        df["finding_labels"].nunique(),
        (df["view_position"] == "PA").sum(),
        (df["view_position"] == "AP").sum()
    ]
})

quality_report
```
